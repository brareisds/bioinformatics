# Busca por 16S rRNA no SILVA

Buscou-se os organismos no SILVA através do nome do organismo no database "rSSU 138.2", um banco de dados focados em pequenas subunidades do RNA. O SILVA mapeia para a sequência que fica no banco de dados ENA (European Nucleotide Archive). Os acessos para o SILVA e ENA são os mesmos.

| Organismo | Ambiente | Filo | Acesso/Link ENA |
|---|---|---|---|
| *Thermocrinis ruber* | Hipertermófilo | Aquificae | [AJ005640](https://www.ebi.ac.uk/ena/browser/view/AJ005640) |
| *Thermotoga maritima* | Hipertermófilo | Thermotogae | [AJ401017](https://www.ebi.ac.uk/ena/browser/view/AJ401017) |
| *Escherichia fergusonii* | Mesófilo | Proteobacteria| [AB681775](https://www.ebi.ac.uk/ena/browser/view/AB681775) |
| *Bacillus subtilis* | Mesófilo | Firmicutes | [AB055007](https://www.ebi.ac.uk/ena/browser/view/AB055007) |
| *Psychrobacter arcticus* | Psicrófilo | Proteobacteria | [AY444823](https://www.ebi.ac.uk/ena/browser/view/AY444823)|
